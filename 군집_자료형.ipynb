**군집 자료형**

여러 데이터를 모은 집합 형태 자료형

1. 리스트(List)
데이터를 순차적으로 저장 -> 열거형

2. 튜플(Tuple)
값을 변경할 수 없는 열거형 집합

3. 세트(Set)
순서가 없고 중복이 허용되지 않는 집합

4. 사전(Dictionary)
키와 값의 쌍으로 구성된 집합

```python
li = ['a', 'b', 'c']
print(li)
```

```python
li[2] = 'd'
print(li[2])
```

```python
li = ['a', 'b', 'c', 'd', 'e']
print('리스트 기본값 출력 : ', li)

print('위치 찾기 : ', li.index('c')) ## 위치 찾기

li.append('f') ## 추가하기1
li.insert(0, 'aa') ## 추가하기2

print('리스트 추가 후 출력 : ', li)

li.remove('aa') ## 삭제하기
del li[2] ## 삭제하기

print('값을 제거 후 출력 : ', li)

print('특정값 확인 : ', 'a' in li)

print('전체 개수 출력 : ', len(li))

print('개수 세기 : ', li.count('a'))
```

```python
num = [1, 3, 2, 5, 4, 7, 9, 8, 6, 10]

print('합 : ', sum(num))
print('최소값 : ', min(num))
print('최대값 : ', max(num))

num.reverse() ## 역순 만들기
print('역순 만든 후 출력 : ', num)

num.sort() ## 오름차순 정렬
print('오름차순 정렬 후 출력 : ', num)

num.sort(reverse=True) ## 내림차순 정렬
print('내림차순 정렬 후 출력 : ', num)
```

# Tuple

강의 : https://www.youtube.com/watch?v=o-Dp7Dm-4KY&list=PLz2iXe7EqJOOTNTK27a4-WsgZU5NVfguh&index=12

```python
# 튜플 만들기

tu = ('a', 'b', 'c')
print(tu)

print(tu[0])

# Error
# tu[2] = 'd'

print('위치 찾기 : ', tu.index('c'))
print('값 존재 여부 확인 : ', 'b' in tu)
print('값 개수 세기 : ', tu.count('a'))
```

```python
num = (5, 6, 9)

# 변수 할당
n1, n2, n3 = num
print(f'변수 할당 후 출력 n1 : {n1}, n2 : {n2}, n3 : {n3}')

# 값 교환하기
a = 'hello'
b = 'world'
print(f'교환 전 값 출력 a : {a}, b : {b}')
(a, b) = (b, a)
print(f'교환 후 값 출력 a : {a}, b : {b}')
```

```python
# 리스트 -> 튜플
li = ['a', 'b', 'c', 'd', 'e', 'f']
li = tuple(li)
print('리스트를 튜플로 변경 후 값 출력 : ', li)

# 튜플 -> 리스트
tu = ('a', 'b', 'c')
tu = list(tu)
print('튜플을 리스트로 변경 후 값 출력 : ', tu)
```

# Set

강의 : https://www.youtube.com/watch?v=ObR6YOlPVno&list=PLz2iXe7EqJOOTNTK27a4-WsgZU5NVfguh&index=13

```python
se = set() # 비어있는 Set
# se = {} -> 딕셔너리
```

```python
# 세트 특징(순서, 중복 없음)
a = {2, 4, 6, 8}
print(a)

b = {2, 4, 2, 1, 2, 3}
print(b)

# 인덱스 X

# 세트 활용
a.add(5) # 추가하기
print('추가 후 출력 : ', a)

a.remove(5) # 삭제하기
print('제거 후 출력 : ', a)

print('값 확인 : ', 1 in a)

print('전체 개수 : ', len(a))
print('합 구하기 : ', sum(a))
print('최소값 찾기 : ', min(a))
print('최대값 찾기', max(a))

# 리스트로 만들기
a = list(a)
print('Set -> List : ', a)

# 튜플 만들기
a = tuple(a)
print('Set(?) -> Tuple : ', a)
```

```python
a = {1, 2, 3}
b = {2, 3, 4}

print('교집합 : ', a&b)
print('합집합 : ', a|b)
print('차집합 : ', a-b)
print('대칭 차집합 : ', a^b)
```

# Dictionary

강의 : https://www.youtube.com/watch?v=azITuZ-7G8M&list=PLz2iXe7EqJOOTNTK27a4-WsgZU5NVfguh&index=14

dic = {키: 값, 키: 값, 키: 값}

```python
# 딕셔너리 만들기
dic = {}
dic = dict()
```

```python
dic = {'kor':80, 'eng':90, 'mat':77}
print("키 값으로 값 출력(국어점수 출력) : ", dic['kor'])
dic['kor'] = 85
dic['sci'] = 92
print('값 변경 및 추가 후 출력 : ', dic)

del dic['mat'] # 삭제하기
print('데이터 삭제 후 전체 리스트 출력 : ', dic)
dic.clear() # 전체 삭제
```

```python
dic = {'kor':80, 'eng':90, 'mat':77}
print('리스트 출력 : ', dic)

print('값 확인하기(키 기준) : ', 'eng' in dic)
print('전체 개수 : ', len(dic))

print('모든 키 얻기 : ', dic.keys())
key_li = list(dic.keys())
print('키값을 리스트로 생성 : ', key_li)

value_li = list(dic.values())
print('value를 리스트로 생성 : ', value_li)

print('모든 순서쌍 얻기 : ', dic.items())
```

```python
li = ['ab', 'cd', 'ef']
print('리스트 출력 : ', li)
li = dict(li)
print('리스트를 딕셔너리로 변경 후 출력 : ', li)
```

```python
li = [['a', 1], ['b', 2], ['c', 3]]
print("리스트 출력 :", li)
li = dict(li)

print('리스트를 딕셔너리로 변경 후 출력 : ', li)
```
