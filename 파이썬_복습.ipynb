# 11강. 모듈

학습 목차
1. 모듈의 이해
2. 유용한 모듈

**math 모듈 멤버**

pi : 원주율

fabs(x) : x의 절대값

ceil(x) : x의 가장 가까운 정수로 올림

floor(x) : x의 가장 가까운 정수로 버림

**random 모듈 멤버**

random() : 0~1 사이의 숫자 중 난수 발생

randint(a, b) : a부터 b 사이의 숫자 중 난수 발생

randrange(a, b, c) : a부터 b 사이의 c의 간격으로 나열된 숫자 중 난수 발생

choice(sequence) : 주어진 항목을 랜덤하게 반환

sample(sequence) : 랜덤하게 여러 개의 원소를 선택

shuffle(sequence) : 시퀀스의 순서를 랜덤하게 섞음

**time 모듈 멤버**

time() : 1970.1.1 자정 이후로 누적된 초를 실수 단위로 반환

gmtile() : struct_time 형식으로 기준 시각 반환

localtime(time) : 입력된 초를 변환하여, 지방표준시 기준 시각으로 변환

strftime(str, time) : 지정된 형식으로 시각을 반환

ctime() : 현재 시간을 반환

sleep(n) : 현재 동작중인 프로세스를 주어진 n초만큼 멈춤

파일 함수

- 파일과 연결되어 있는 파일 객체 생성

read()
- 특정 개수의 문자를 반환

readline()
- 한 라인의 문자열을 반환

readlines(): list
- 전체 라인의 문자열을 리스트로 반환

**파일 객체 생성**

```
파일객체_참조변수 = open(파일이름, 모드)
```
'r' : 읽기 용도

'w' : 새로운 파일을 쓰기 용도

'a' : 파일의 끝에 데이터를 덧붙이기 용도

````python
students = [
    (20251234, '김철수', '컴퓨터과학', 2, 3.8),
    (20265678, '이영희', '생활과학부', 3, 4.2),
    (20233456, '박민수', '사회복지학과', 1, 3.5)
]
print('(명시) 이름과 성적 출력')
for stn, name, deft, year, grade in students:
    print(f'{name}의 성적 : {grade}')

print('(언더스코어) 이름과 성적 출력')
for _, name, _, _, grade in students:
    print(f'{name}의 성적 : {grade}')

# print('파이썬의 한계(?)')
# for name, grade in students:
#     print(f'{name}의 성적 : {grade}')
````
